## Nível 1 (Básico): Os operadores

**Objetivo da aula:** Usar operadores (`+ - * / // % **`) e funções matemáticas (`math.sqrt`, `math.pi`, `round`) em problemas técnicos.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
a = 17                               # primeiro número
b = 5                                # segundo número
print("Soma:", a + b)                # 22
print("Subtração:", a - b)           # 12
print("Multiplicação:", a * b)       # 85
print("Divisão:", a / b)             # 3.4
print("Divisão inteira:", a // b)    # 3 (quantas vezes o 5 cabe inteiro no 17)
print("Resto:", a % b)               # 2 (o que sobra da divisão inteira)
print("Potência:", a ** 2)           # 289 (17 ao quadrado)
```
