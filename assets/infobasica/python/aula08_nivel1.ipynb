## Nível 1 (Básico): Verdadeiro ou falso

**Objetivo da aula:** Construir expressões booleanas com `== != < > <= >=` e `and`, `or`, `not`, e montar tabelas-verdade.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
print(5 > 3)        # True: 5 é maior que 3
print(5 == 3)       # False: == compara se os dois são iguais
print(5 != 3)       # True: != significa "diferente"
nota = 8            # um valor de exemplo
print(nota >= 7)    # True: 8 é maior ou igual a 7
```
