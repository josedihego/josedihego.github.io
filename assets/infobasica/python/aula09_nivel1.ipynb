# Aula 9 – Decisão simples e composta: if / else

## Nível 1 (Básico): Par ou ímpar

**Objetivo da aula:** Usar `if`/`else` para tomar decisões e reproduzir a mesma lógica no Scratch e na planilha com `SE`.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
numero = int(input("Número inteiro: "))   # lê o número
if numero % 2 == 0:                       # resto zero na divisão por 2?
    print("Par")                          # bloco do "se" (repare no recuo)
else:                                     # senão...
    print("Ímpar")                        # bloco do "senão"
```
