# Aula 10 – Decisões encadeadas: if / elif / else

## Nível 1 (Básico): Ácido, neutro ou básico

**Objetivo da aula:** Tratar várias faixas de valores com `elif` e com `SE` aninhado ou `SES` na planilha.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
ph = float(input("pH da solução: "))   # lê o pH
if ph < 7:                             # abaixo de 7
    print("Ácido")                     # primeira faixa
elif ph == 7:                          # elif: só é testado se o if acima foi falso
    print("Neutro")                    # exatamente 7
else:                                  # sobrou: acima de 7
    print("Básico")                    # última faixa
```
