# Aula 13 – Introdução à repetição: for e range

## Nível 3 (Avançado): Tabuada

**Objetivo da aula:** Repetir instruções com `for` e `range(início, fim, passo)` e relacionar com `repita` no Scratch e com séries na planilha.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
numero = int(input("Tabuada do: "))                         # lê o número
for multiplicador in range(1, 11):                          # de 1 a 10
    resultado = numero * multiplicador                      # calcula a linha da tabuada
    print(f"{numero} x {multiplicador} = {resultado}")      # mostra no formato 7 x 3 = 21
```
