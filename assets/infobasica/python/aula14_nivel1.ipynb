# Aula 14 – Acumuladores: somas e médias

## Nível 1 (Básico): Soma de 1 a 100

**Objetivo da aula:** Usar uma variável acumuladora dentro do laço para calcular somas e médias.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
soma = 0                          # acumulador começa em zero
for numero in range(1, 101):      # de 1 a 100
    soma = soma + numero          # acrescenta o número ao total
print("Soma de 1 a 100:", soma)   # 5050
```
