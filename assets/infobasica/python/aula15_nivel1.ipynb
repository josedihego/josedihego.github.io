# Aula 15 – Contadores, maior e menor

## Nível 1 (Básico): Contar os pares

**Objetivo da aula:** Combinar laços e decisões para contar ocorrências e encontrar o maior e o menor valor.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
pares = 0                           # contador começa em zero
for numero in range(1, 51):         # de 1 a 50
    if numero % 2 == 0:             # é par?
        pares = pares + 1           # conta mais um
print("Pares entre 1 e 50:", pares) # 25
```
