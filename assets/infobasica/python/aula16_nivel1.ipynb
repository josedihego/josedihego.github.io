# Aula 16 – Repetição com condição: while

## Nível 1 (Básico): Contar com while

**Objetivo da aula:** Usar `while` quando o número de repetições não é conhecido e evitar laços infinitos.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
contador = 1                    # começa em 1
while contador <= 5:            # repete ENQUANTO a condição for verdadeira
    print(contador)             # mostra o valor atual
    contador = contador + 1     # sem esta linha o laço nunca terminaria!
print("Fim")                    # executa quando a condição fica falsa
```
