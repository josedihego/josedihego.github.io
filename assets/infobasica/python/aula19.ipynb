# Aula 19 – Laços com decisões: múltiplos, divisores e sequências

**Objetivo:** Resolver problemas que combinam `for`, `%` e contadores, e usar `break`.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
for numero in range(1, 101):                # percorre de 1 a 100
    if numero % 3 == 0 or numero % 7 == 0:  # resto zero = divisível
        print(numero, end=" ")              # mostra na mesma linha
print()                                     # quebra a linha
anterior = 0                                # primeiro termo de Fibonacci
atual = 1                                   # segundo termo
for posicao in range(20):                   # 20 termos
    print(anterior, end=" ")                # mostra o termo atual
    proximo = anterior + atual              # cada termo é a soma dos dois anteriores
    anterior = atual                        # avança uma posição
    atual = proximo                         # avança uma posição
    if anterior > 1000:                     # passou de 1000?
        break                               # break sai do laço na hora
print()                                     # quebra a linha
```
