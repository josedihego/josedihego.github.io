# Aula 19 – Laços com decisões: múltiplos, divisores e sequências

## Nível 1 (Básico): Múltiplos de 3 ou de 7

**Objetivo da aula:** Resolver problemas que combinam `for`, `%` e contadores, e usar `break`.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
for numero in range(1, 101):                # percorre de 1 a 100
    if numero % 3 == 0 or numero % 7 == 0:  # resto zero = divisível
        print(numero, end=" ")              # end=" " continua na mesma linha
print()                                     # quebra a linha no final
```
