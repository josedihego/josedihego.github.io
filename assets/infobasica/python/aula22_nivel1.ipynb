# Aula 22 – Simulações passo a passo com laços

## Nível 1 (Básico): Bactérias que dobram

**Objetivo da aula:** Simular fenômenos que mudam com o tempo, atualizando valores a cada passo de um laço, e comparar com a planilha linha a linha.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
bacterias = 100                                     # população inicial
for hora in range(1, 11):                           # 10 horas
    bacterias = bacterias * 2                       # a cada hora a população dobra
    print(f"Hora {hora}: {bacterias} bactérias")    # mostra cada passo
```
