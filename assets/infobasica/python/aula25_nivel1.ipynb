# Aula 25 – Textos (strings)

## Nível 1 (Básico): Tamanho e maiúsculas

**Objetivo da aula:** Manipular textos: leitura, tamanho, maiúsculas/minúsculas, fatias e formatação.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
palavra = input("Palavra: ")              # lê o texto
print("Letras:", len(palavra))            # len: quantidade de caracteres
print("Maiúsculas:", palavra.upper())     # upper: tudo em maiúsculas
print("Minúsculas:", palavra.lower())     # lower: tudo em minúsculas
print("Primeira letra:", palavra[0])      # posição 0 = primeiro caractere
```
