# Aula 26 – Listas: criação, índices e leitura

## Nível 1 (Básico): Uma lista de cores

**Objetivo da aula:** Guardar vários valores numa lista, acessar por índice e preenchê-la com um laço.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
cores = ["vermelho", "verde", "azul"]   # lista com 3 textos, entre colchetes
print(cores[0])                         # índice 0 = primeiro elemento
print(cores[-1])                        # índice -1 = último elemento
print(len(cores))                       # quantidade de elementos: 3
cores.append("amarelo")                 # append acrescenta no fim
print(cores)                            # mostra a lista inteira
```
