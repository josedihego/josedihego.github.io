# Aula 27 – Percorrendo listas: soma, média, maior e menor

## Nível 1 (Básico): Funções prontas de listas

**Objetivo da aula:** Percorrer listas com `for` para calcular estatísticas e comparar com `sum`, `max`, `min` e `sorted`.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
notas = [7.5, 8.0, 6.5, 9.0]                 # lista de notas
print("Soma:", sum(notas))                   # sum soma todos os elementos
print("Média:", sum(notas) / len(notas))     # soma ÷ quantidade
print("Maior:", max(notas))                  # max: maior elemento
print("Menor:", min(notas))                  # min: menor elemento
print("Ordenadas:", sorted(notas))           # sorted: nova lista em ordem crescente
```
