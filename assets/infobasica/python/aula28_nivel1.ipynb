# Aula 28 – Busca em listas

## Nível 1 (Básico): Está na lista?

**Objetivo da aula:** Implementar a busca sequencial e usar o operador `in`, `index()` e `PROCV`/`CORRESP` na planilha.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
frutas = ["banana", "maçã", "manga", "caju"]   # lista de frutas
fruta = input("Fruta: ")                       # o que procurar
if fruta in frutas:                            # in: está na lista?
    print("Tem", fruta)   # encontrou
else:   # caso contrário
    print("Não tem", fruta)   # não encontrou
```
