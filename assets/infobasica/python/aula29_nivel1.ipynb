# Aula 29 – Filtrando e transformando listas

## Nível 1 (Básico): Só os pares

**Objetivo da aula:** Criar novas listas a partir de critérios e transformações, com laço + `if` e com compreensão de listas.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
numeros = [3, 8, 12, 5, 20, 7]     # lista original
pares = []                         # lista nova, vazia
for numero in numeros:             # percorre a original
    if numero % 2 == 0:            # é par?
        pares.append(numero)       # guarda na nova
print(pares)                       # [8, 12, 20]
```
