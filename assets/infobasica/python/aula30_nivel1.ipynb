# Aula 30 – Funções: def, parâmetros e retorno

## Nível 1 (Básico): Minha primeira função

**Objetivo da aula:** Organizar o código em funções com parâmetros e `return`, e reutilizar funções dentro de outras.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
def saudacao(nome):              # def cria a função; nome é o parâmetro
    print("Olá,", nome)          # o que a função faz
saudacao("Ana")                  # chama a função passando "Ana"
saudacao("Caio")                 # a mesma função, outro valor
```
