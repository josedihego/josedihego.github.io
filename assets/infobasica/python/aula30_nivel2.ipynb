## Nível 2 (Intermediário): Função que devolve valor

**Objetivo da aula:** Organizar o código em funções com parâmetros e `return`, e reutilizar funções dentro de outras.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
def celsius_para_fahrenheit(celsius):   # recebe a temperatura em °C
    return celsius * 9 / 5 + 32         # return devolve o resultado para quem chamou
for c in range(0, 101, 20):             # 0, 20, 40, 60, 80, 100
    f = celsius_para_fahrenheit(c)      # guarda o valor devolvido
    print(c, "°C =", f, "°F")           # mostra a tabela
```
