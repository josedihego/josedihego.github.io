# Aula 32 – Dicionários: dados com nome

## Nível 1 (Básico): Dicionário de capitais

**Objetivo da aula:** Guardar pares chave-valor em dicionários e agrupar dados por categoria.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
capitais = {"BA": "Salvador", "PE": "Recife", "SP": "São Paulo"}   # chave: valor
print(capitais["BA"])                         # acesso pela chave: Salvador
capitais["MG"] = "Belo Horizonte"             # chave nova: acrescenta
print(len(capitais))                          # 4 pares
for sigla, capital in capitais.items():       # items: percorre os pares
    print(sigla, "→", capital)   # BA → Salvador ...
```
