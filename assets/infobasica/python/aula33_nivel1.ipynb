## Nível 1 (Básico): Lista de tarefas

**Objetivo da aula:** Integrar listas, dicionários, funções e menu num sistema completo.

Leia os comentários (depois do `#`) e execute a célula abaixo clicando em ▶. Depois altere os valores e execute de novo.

```python
tarefas = []                                               # lista vazia
while True:                                                # repete até o break
    tarefa = input("Nova tarefa (em branco para terminar): ")   # lê uma tarefa
    if tarefa == "":                                       # nada digitado?
        break                                              # termina
    tarefas.append(tarefa)                                 # guarda
for numero, tarefa in enumerate(tarefas, start=1):         # enumerate numera a partir de 1
    print(numero, "-", tarefa)   # 1 - Estudar, 2 - Ler ...
```
